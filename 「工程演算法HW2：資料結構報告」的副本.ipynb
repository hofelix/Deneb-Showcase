## 步驟一：迴圈與遞迴的對照程式碼

**1. 指令要求：**
使用 Python 分別以 `for/while loop` 以及 `Recursion`（遞迴）寫出計算費氏數列（Fibonacci）第 $n$ 項的程式碼。

**2. AI 生成程式碼與標註：**

**【版本 A：迴圈實作 (Loop)】**
迴圈版本適合用在已知確切迴圈數量（如 `for loop`）的情境。

```
def fibonacci_loop(n):
    # 處理邊界條件
    if n == 0: return 0
    if n == 1: return 1

    a, b = 0, 1
    # 已知迴圈次數，從第 2 項計算到第 n 項
    for i in range(2, n + 1):
        a, b = b, a + b
    return b

print(f"迴圈版本第 5 項: {fibonacci_loop(5)}")
```

**【版本 B：遞迴實作 (Recursion)】**
遞迴的核心在於函式會不斷呼叫自己，直到滿足終止條件才開始回傳。

```
def fibonacci_recursion(n):
    # 終止條件：避免無窮迴圈
    if n == 0: return 0
    if n == 1: return 1

    # 📌 關鍵行：在此處自己呼叫自己 (Self-calling)
    # 將問題拆解為前兩項的總和
    return fibonacci_recursion(n - 1) + fibonacci_recursion(n - 2)

print(f"遞迴版本第 5 項: {fibonacci_recursion(5)}")
```

## 步驟二：DFS 與 BFS 策略的程式碼

**1. 指令要求：**
實作 DFS（深度優先搜尋）與 BFS（廣度優先搜尋）演算法來遍歷一個簡單的圖結構（模擬講義中「找橘子」的迷宮案例），並說明順序差異與資料容器。

**2. AI 生成程式碼與觀察：**

我們定義一個簡單的樹狀圖，假設目標「橘子」放在節點 `E`：

```
# 圖結構 (鄰接串列)
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'], # 假設 E 是橘子
    'C': ['F'],
    'D': [], 'E': [], 'F': []
}
```

**【策略 A：DFS 深度優先搜尋】**

* **資料容器觀察：** DFS 通常利用系統的 **Call Stack (呼叫堆疊)** 來達成遞迴。

```
def dfs_find_orange(graph, node, target, visited=None):
    if visited is None: visited = set()

    if node not in visited:
        print(f"造訪節點: {node}")
        visited.add(node)

        # 找到目標，準備觸發 Return
        if node == target:
            print(">> 找到橘子了！(準備回頭)")
            return True

        # 針對鄰居「走到底」
        for neighbor in graph[node]:
            if dfs_find_orange(graph, neighbor, target, visited):
                return True # 只要其中一條路找到，就一路 Return 回去
    return False

print("--- DFS 搜尋順序 ---")
dfs_find_orange(graph, 'A', 'E')
```

**【策略 B：BFS 廣度優先搜尋】**

* **資料容器觀察：** BFS 必須使用 **Queue (佇列)**，具有「先進先出 (FIFO)」的特性。

```
from collections import deque

def bfs_find_orange(graph, start, target):
    queue = deque([start]) # 使用 Queue 資料容器
    visited = set([start])

    while queue:
        node = queue.popleft() # 從佇列最前方取出
        print(f"造訪節點: {node}")

        if node == target:
            print(">> 找到橘子了！")
            return True

        # 將下一層的所有鄰居加入 Queue，準備「平均走」
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    return False

print("--- BFS 搜尋順序 ---")
bfs_find_orange(graph, 'A', 'E')
```

**順序差異說明：**

* **DFS 尋找順序**：`A -> B -> D -> E`。它會選擇一條分支（往 B、D）一直深入到底，發現 D 沒有路且不是橘子後，才**回溯 (Backtrack)** 去找 B 的另一個分支 E。
* **BFS 尋找順序**：`A -> B -> C -> D -> E`。它會先看完距離 A 第一層的所有節點 (B, C)，確認沒有橘子後，再「平均地」去看下一層 (D, E, F)，因此會在第二層找到 E。

## 步驟三：本質分析與講義概念體現

根據 Lecture #2 與 Lecture #3 的內容，分析上述程式碼背後的工程本質：

**1. 遞迴中的 Return 體現 (Backtracking 的本質)**
在 DFS 與遞迴程式碼中，`return` 扮演了極為關鍵的控制角色。誠如講義 Lecture #3 提到的「找到橘子，整個路線就直接回頭」。

* 當程式碼執行到 `if node == target: return True` 時，它不僅僅是回傳一個布林值，更代表這條 Call Stack 的深入探索正式終結。
* 系統會停止往下探索，並沿著原路（上一層的函式呼叫）一路 `return` 回到起點。這就是演算法中「回溯 (Backtracking)」機制的具體程式碼體現。

**2. BFS 中的「平均走」體現與 Queue 的運用**
講義中提到 BFS 是一種「平均走」的策略，適合尋找最短路徑。

* 程式碼中體現為 `queue = deque()` 與 `queue.popleft()`。
* 當程式在節點 A 時，會將 B、C 放進 Queue 的尾端。由於 Queue 是**先進先出 (FIFO)**，程式必定會先處理完 B 與 C（第一層），才會去處理 B 與 C 延伸出來的 D、E、F（第二層）。這種資料結構的設計強迫演算法像同心圓一樣向外擴展，完美實踐了「平均探測每一條路線」的概念。

**3. DFS 的「走到底」與找第一顆橘子**
講義強調 DFS 適合「快速找出合格解（不用管是不是最大的，找到第一顆就好）」。

* 程式碼中 `dfs_find_orange` 使用了 `for neighbor in graph[node]: dfs(...)`。
* 它不會管另一邊（C 節點）有多近，只要 B 節點有路，它就會毫不猶豫地往 D 衝到底（Depth-First）。
* 在尋找單一目標（如迷宮出口、找任一橘子）時，只要運氣好選對分支，DFS 的寫法通常更簡潔，且能極快地找到第一個合格的答案。
